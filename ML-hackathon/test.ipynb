```python
def vanligaste(filename):
    dic = {}
    with open (filename) as file:
        for line in file:
            for bok in line.strip().split():
                if bok in dic:
                    dic[bok] += 1
                else:
                    dic[bok] = 1
    antal = 0
    for name, value in dic.items():
        if antal == 5:
            break
        else:
            print(name, value)
            antal += 1

vanligaste("monster_random.txt")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

img = data.camera()                  # en gråskalebild som NumPy-array
print(img.shape, img.dtype, img.min(), img.max())   # (512, 512) uint8 0 255
print(img[100, 200])                 # pixelvärdet på rad 100, kolumn 200 (54)

crop = img[100:200, 150:300]         # rader 100-199, kolumner 150-299
mask = img > 128                     # True där pixeln är ljus
print(mask.mean())                   # andel ljusa pixlar, ca 0.64

fig, ax = plt.subplots(1, 3, figsize=(10, 3))
ax[0].imshow(img, cmap='gray')
ax[1].hist(img.ravel(), bins=50)
ax[2].imshow(mask, cmap='gray')
plt.show()
```
