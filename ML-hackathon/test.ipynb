```python
def vanligaste(filename):
    dic = {}
    with open (filename) as file:
        for line in file:
            for bok in line.strip().split():
                if bok in dic:
                    dic[bok] += 1
                else:
                    dic[bok] = 1
    antal = 0
    for name, value in dic.items():
        if antal == 5:
            break
        else:
            print(name, value)
            antal += 1

vanligaste("monster_random.txt")
```
